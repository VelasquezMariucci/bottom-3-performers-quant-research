When going over selected key metrics in the S&P DOW JONES Selected Key Metrics Dashboard, talk about:
- Dividend Yield and how it relates to bonds.
- How a high P/E ratio can either mean A high P/E ratio often means a stock's price is expensive and investors expect high future growth. A low P/E can indicate an undervalued stock, or that the company has struggled recently. (Since its how myuch investors are willing to pay for each dollar of  earnings).
- How a P/B ratio under 1.0 generally suggests a stock is undervalued, while a higher ratio indicates investors are willing to pay a premium for growth.
- In P/S, looking at the revenue alongside the cost of production is important, since we need the profits. This metric shows how investors value the company sales (they buy if they think sales are going to increase and sell if they are going to decrease).
- If a Beta's absolute value is < 1 or beta < 0, it is more sensitive to broader market swings than average. If Beta is between -0.5 and +0.5, its price movements are largely independent of broad market swings (remember this is a correlation metric, not causation).
- CONSTITUENT VOLATILITY is individual volatility.
- RSI - 30 DAY (Relative Strength Index) shows the average gain over the average loss in the last 30 days. 70% overbought, 30 oversold, 50 ok.
- Value scores look at cheap fundamentals. A positive score means you are getting more assets, earnings, or sales for every dollar you invest compared to the rest of the market.
-  Growth scores measure expansion momentum. A positive score means these companies are growing their revenues, profits, and sales faster than the market average

For the last two:
Positive Scores (> 0): This indicates that the sector has a stronger bias or "tilt" toward that factor than the broader market benchmark.
Negative Scores (< 0): This means the sector underperforms or lacks that specific characteristic relative to the market average.
